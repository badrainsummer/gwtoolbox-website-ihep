```python
# setting the path to the toolbox
import sys
#sys.path.insert(1,'/home/raojinhui/gwtoolbox/gwtoolbox/')

# importing the needed module
import numpy as np
from astropy.cosmology import Planck18 as cosmos
from gwtoolbox.tools_earth import Tools
from gwtoolbox.tools_EMC import *
from gwtoolbox.EM_HE import *
import matplotlib.pyplot as plt

%matplotlib inline
pop_para=[100,10,1.4,0.2,1.1,2.5,0.1]
```

```python
# Search for GRB
from gwtoolbox.HE_models import *
# parameter tuning!
pi=3.1415926
half_open_mean=5.*pi/180.
half_open_std=1.*pi/180.
lg_r_grb_mean=14.5
lg_r_grb_std=0.1
lgEgrb_mean=49.3
lgEgrb_std=0.5
nup0_mean=0.003
nup0_std=0.001
jet_Gamma_mean=250
jet_Gamma_std=30

Band_alpha=-1.1
Band_beta=-2.5
Band_s=2
t_smear_intrinsic=0.3

model_para=[half_open_mean,half_open_std, lg_r_grb_mean, lg_r_grb_std,lgEgrb_mean, lgEgrb_std,nup0_mean, nup0_std, jet_Gamma_mean, jet_Gamma_std]

detector_type='ligo'
event_type='nsns'
population='I'
rho_cri=8. #snr
cosmos=cosmos
new_theta=pop_para
T=24.*60.*3650 # 2 years,1 month=43829.1 min

HE_detector='Fermi_GBM'
signal_type='sGRB_p'
```

```python
tool=Tools(detector_type=detector_type, event_type=event_type, population=population,cosmos=cosmos, new_theta=pop_para)

total=tool.total_number(time_obs=T, rho_cri=rho_cri) #rho_cri? SNR?
print("expected number is: %f" % total)
total_poisson=np.random.poisson(lam=total)
sensitivity = tool.sensitivity_curve()
print(total_poisson,type(total_poisson))
```
